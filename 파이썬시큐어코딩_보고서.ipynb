**컴퓨터와 1대 1주사위 게임**

규칙
- 1. 참여자는 이름을 작성한다.
- 2. 참여자는 단판부터 5판까지 승리 조건을 입력한다.
- 3. 컴퓨터부터 2개의 주사위를 굴리고, 이후 참여자 또한 주사위를 굴린다.각자 주사위의 합이 더 높을 쪽으로 승률을 쌓는다.
- 4. 승리 조건 이전까지 3번 순서를 반복한다.
- 5. 승리 조건 달성 시 반복문이 끝나며 "ㅁㅁ의 n판m선 승!"이라는 출력과 함께 게임이 끝나게 된다.

```python
import random as r

def Stage_role():
    while True:
        try:
            stage_number_input = input("입력하시는 판의 수가 승리 조건을 결정합니다. (1부터 5까지) 정수로 입력해주십시오. :")
            stage_number = int(stage_number_input)

            if 1 <= stage_number <= 5:
                required_wins = (stage_number // 2) + 1
                return stage_number, required_wins
            else:
                print("승리 조건은 1부터 5까지의 정수로만 입력할 수 있습니다. 다시 입력해주세요.")
        except ValueError:
            print("유효한 정수를 입력해주십시오. 문자는 입력할 수 없습니다.")
        except Exception as e:
            print(f"오류 발생: {e}, 다시 시도해주세요.")


def DiceGame():
    # 1. 참여자 이름 설정
    while True:
        PlayerName = input("참여자의 이름을 입력하십시오: ").strip()
        if PlayerName:
            break
        else:
            print("이름을 꼭 입력해야 합니다. 다시 입력해주세요.")

    # 2. 승리 조건 입력 및 설정
    total_stages, required_wins = Stage_role()
    print(f"\n{PlayerName}님, 총 {total_stages}판 중 {required_wins}선승을 먼저 달성하면 승리합니다!")

    ComputerScore = 0
    PlayerScore = 0
    current_stage = 1

    # 4. 승리 조건 이전까지 순서를 반복한다.
    while current_stage <= total_stages:
        print(f"\n----- {current_stage}번째 라운드 시작! -----")

        print("컴퓨터가 주사위를 굴립니다...")
        ComputerDice = [r.randint(1, 6), r.randint(1, 6)]
        computer_sum = sum(ComputerDice)
        print(f"컴퓨터: 주사위 눈금 {ComputerDice[0]}, {ComputerDice[1]} (합계: {computer_sum})")

        input(f"{PlayerName}님, 주사위를 굴리려면 Enter 키를 누르세요...")
        PlayerDice = [r.randint(1, 6), r.randint(1, 6)]
        player_sum = sum(PlayerDice)
        print(f"{PlayerName}: 주사위 눈금 {PlayerDice[0]}, {PlayerDice[1]} (합계: {player_sum})")

        if computer_sum > player_sum:
            print("이번 라운드는 컴퓨터 승리!")
            ComputerScore += 1
        elif player_sum > computer_sum:
            print(f"이번 라운드는 {PlayerName} 승리!")
            PlayerScore += 1
        else:
            print("이번 라운드는 무승부입니다!")

        print(f"현재 스코어: {PlayerName} {PlayerScore} : {ComputerScore} 컴퓨터")

        if PlayerScore == required_wins:
            print(f"\n----- 게임 종료! -----")
            print(f"{PlayerName}의 {total_stages}판 {required_wins}선승 승리!")
            break
        elif ComputerScore == required_wins:
            print(f"\n----- 게임 종료! -----")
            print(f"컴퓨터의 {total_stages}판 {required_wins}선승 승리!")
            break

        current_stage += 1

    else:
        print("\n----- 모든 라운드 종료! 최종 결과는 다음과 같습니다. -----")
        if PlayerScore > ComputerScore:
            print(f"{PlayerName}이(가) 더 많은 라운드를 이겼습니다! 최종 스코어: {PlayerName} {PlayerScore} : {ComputerScore} 컴퓨터")
        elif ComputerScore > PlayerScore:
            print(f"컴퓨터가 더 많은 라운드를 이겼습니다! 최종 스코어: {PlayerName} {PlayerScore} : {ComputerScore} 컴퓨터")
        else:
            print(f"모든 라운드를 진행했지만, 최종 스코어는 {PlayerScore} : {ComputerScore}로 무승부입니다!")

if __name__ == "__main__":
    DiceGame()
```

-- 아래 --

컴퓨터와 가위바위보 대결을 합니다. 선택을 숫자로 해주세요(가위:0, 바위:1, 보:2) : 2

당신은 보를 선택했습니다.

컴퓨터는 바위를 선택했습니다.

당신이 이겼습니다



```python
import random as r
player = int(input("컴퓨터와 가위바위보 대결을 합니다. 선택을 숫자로 해주세요(가위:0, 바위:1, 보:2) : "))
computer = r.randint(0,2)
choice = ["가위", "바위","보"]

print(f"\n당신은 {choice[player]}를 선택했습니다.")
print(f"컴퓨터는 {choice[computer]}를 선택했습니다.")

result = player - computer
if result == 0:
    print("\n비겼습니다.")
elif result in [-2, 1]:
    print("\n당신이 이겼습니다.")
else:
    print("\n컴퓨터가 이겼습니다.")
```

11. 아래의 코드를 완성하여 2부터 100까지의 정수중 솟수(Prime Number)들의 합을 구하고 출력할 수 있도록 코드를 완성하여 제출합시다(솟수는 1과 자기자신 외에는 나누어떨어지지 않는 수).

*ChatGPT를 활용하여 코드를 얻고, 이를 분석하여 이해하는 방향으로 진행하면 좋습니다.

-- 아래 --

print("2부터 100까지의 수 중 솟수의 합은 %s 입니다" % 합)

```python
sum = 0

for k in range(2, 101):
    for i in range(2, int(k ** 0.5) + 1):
        if k % i == 0:
            break
    else:
        sum += k

print("2부터 100까지의 수 중 소수의 합은 %s 입니다" %sum)
```

12. 주어진 파이썬 코드에 내용을 추가하여  <그림>과 같이 칸을 맞추어 결과를 출력하도록 코드를 완성하여 제출합시다.

- 응시자 = ["cheolsu", "younghee", "ayoung"]

    - 영어   = [90,  100, 80]
    - 평균   = [0,0,0]

```python
students = ["cheolsu", "younghee", "ayoung"]
korean = [100, 100, 97]
english = [90, 80, 80]

print("NAME        KOR  ENG  SUM    AVG")
print("---------------------------------")

for n, k, e in zip(students, korean, english):
    sum = k + e
    ave = sum / 2
    print(f"{n:<10}{k:>5}{e:>5}{sum:>6}{ave:>6.2f}")
```
